```python
import pandas as pd

```

```python
# Define the file names (assuming they are in the same folder as your notebook)
TRAIN_FILE = 'supercars_train.csv'
TEST_FILE = 'supercars_test.csv'

# Load the datasets into pandas DataFrames
df_train = pd.read_csv(TRAIN_FILE)
df_test = pd.read_csv(TEST_FILE)

print("✅ Data Loading Successful!")
print(f"Training Data Shape: {df_train.shape}")

# Display the first 5 rows to verify loading
print("\n--- Training Data Head ---")
df_train.head()
```

```python
# Assuming df_train and df_test are already loaded from the previous step

print("--- Checking and Removing Duplicates from Training Data ---")

# 1. Define columns to check for duplicates
# Exclude 'id' (unique identifier) and 'price' (target variable)
duplicate_check_cols = df_train.columns.drop(['id', 'price'])

# 2. Count the number of duplicate rows
# The 'duplicated' method checks for rows that are identical to a previous row.
duplicate_rows_count = df_train.duplicated(subset=duplicate_check_cols).sum()

print(f"Total number of duplicate rows (based on features) found: {duplicate_rows_count}")

# 3. Conditional Removal of Duplicates
if duplicate_rows_count > 0:
    initial_train_shape = df_train.shape[0]
    
    # Remove duplicates, keeping the first instance found
    df_train = df_train.drop_duplicates(subset=duplicate_check_cols, keep='first')
    
    print(f"✅ Removed {initial_train_shape - df_train.shape[0]} duplicate rows.")
    print(f"New Training Data Shape: {df_train.shape}")
else:
    print("✅ No duplicate rows found in the training set. Data is clean.")

# 4. Check Test Data (Informational Only)
# We check the test set but DO NOT remove duplicates, 
# as every ID in the test set needs a prediction for the competition.
test_duplicate_cols = df_test.columns.drop('id')
test_duplicate_rows_count = df_test.duplicated(subset=test_duplicate_cols).sum()
print(f"\nDuplicates found in Test Set (No Removal): {test_duplicate_rows_count}")
```
